### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
type(tienda['Fecha de Compra'])
```

#1. Análisis de facturación


```python
t1_ingreso_total = tienda['Precio'].sum()
t2_ingreso_total = tienda2['Precio'].sum()
t3_ingreso_total = tienda3['Precio'].sum()
t4_ingreso_total = tienda4['Precio'].sum()
```

```python
print(f'Ingresos totales de cada tienda: \n Tienda 1: {t1_ingreso_total}\n Tienda 2: {t2_ingreso_total}\n Tienda 3: {t3_ingreso_total}\n Tienda 4: {t4_ingreso_total} ')
```

# 2. Ventas por categoría

# 6. Gráficos

```python
import matplotlib.pyplot as plt
# Variables para los gráficos
tiendas_nombres = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

ingresos = [t1_ingreso_total, t2_ingreso_total, t3_ingreso_total, t4_ingreso_total]

calificaciones = [t1_calificacion, t2_calificacion, t3_calificacion, t4_calificacion]

envios_promedio = [t1_envio_promedio, t2_envio_promedio, t3_envio_promedio, t4_envio_promedio]
```

```python
# Ingresos Totales por Tienda (Gráfico de Barras)

# Formateamos los valores de los ingresos para que don José lo entienda fácilmente
formatear_ingresos = []
for ingreso in ingresos:
  formatear_ingresos.append(ingreso / 1000000)

#Diseñamos el gráfico
plt.figure(figsize=(12, 8))
plt.bar(tiendas_nombres, formatear_ingresos, color=['lightgreen', 'lightgreen', 'lightgreen', 'salmon']) # Crea las barras
plt.title('La Tienda 4 tiene los ingresos más bajos', color= 'salmon', fontsize=17.6, fontweight='bold')
plt.ylabel('Ingresos (millones de pesos)', fontsize=12)

plt.grid(axis='y', linestyle='-', alpha=0.7) # Rejilla lineal en el eje Y

plt.show()
```

```python
# Calificación Promedio (Gráfico de Barras Horizontales)

#Diseñamos el gráfico
plt.figure(figsize=(12, 8))
plt.barh(tiendas_nombres, calificaciones, color='lightgreen')
plt.title('Todas las tiendas tienen una calificación promedio similar', fontsize=20, color='green', fontweight='bold')
plt.xlabel('Calificación Promedio (de 1 a 5)', fontsize=12)
plt.ylabel('Tienda', fontsize=12)

# límite de 0 a 5 como números enteros
plt.xlim(0, 5)

# Invertimos la Y para que salga la tienda 1 primero
plt.gca().invert_yaxis()

plt.grid(axis='x', linestyle='-', alpha=0.7) # Rejilla lineal en el eje X

plt.show()
```

```python
# Costo de Envío Promedio (Gráfico de Línea)

#Diseñamos el gráfico
plt.figure(figsize=(12, 8))
plt.plot(tiendas_nombres, envios_promedio, color='salmon', marker='o') # el marker agrega las bolitas en la linea roja
plt.title('La tienda 4 tiene el envío promedio más económico', fontsize=20, color='salmon', fontweight='bold')
plt.xlabel('Tienda', fontsize=12)
plt.ylabel('Costo de Envío (pesos)', fontsize=12)

plt.grid(linestyle='--', alpha=1)
plt.show()
```
